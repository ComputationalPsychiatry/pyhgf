# Ajouter des observations en plusieurs fois

Choisir le noyau **Python (pyhgf)** et exécuter les cellules dans l’ordre.

## 1. Créer un réseau

```python
import numpy as np

from pyhgf.model import Network

hgf = Network().add_nodes().add_nodes(value_children=0)
```

## 2. Donner des observations et tracer les résultats

```python
observations = np.array([0.2, 0.4, 0.3, 0.5, 0.4, 0.6, 0.5, 0.7])

hgf.input_data(observations);
```

```python
hgf.plot_trajectories(figsize=(10, 5));
```

## 3. Reprendre avec de nouvelles observations

On garde le même réseau. Il reprend depuis son dernier état et ajoute les nouvelles observations à l’historique. Le graphique contient maintenant les **16 observations**.

```python
nouvelles_observations = np.array([0.8, 0.9, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2])

hgf.input_data(nouvelles_observations);
```

```python
hgf.plot_trajectories(figsize=(10, 5));
```
